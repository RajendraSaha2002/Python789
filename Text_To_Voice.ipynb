```python
import pyttsx3
import ipywidgets as widgets
from IPython.display import display
```

```python
engine = pyttsx3.init()

# Optional: set default properties
engine.setProperty('rate', 150)
engine.setProperty('volume', 1.0)
```

```python
def speak_text(text, speed):
    if not text.strip():
        print("⚠️ I love you ")
        return

    engine.setProperty("rate", speed)
    engine.say(text)
    engine.runAndWait()
```

```python
text_input = widgets.Text(
    value="Hello Python",
    description="Text:"
)

speed_slider = widgets.IntSlider(
    value=150,
    min=100,
    max=300,
    step=10,
    description="Speed:"
)

button = widgets.Button(description="Speak 🔊")

def on_click(b):
    speak_text(text_input.value, speed_slider.value)

button.on_click(on_click)

display(text_input, speed_slider, button)
```
