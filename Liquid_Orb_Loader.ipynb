```python
import matplotlib.pyplot as plt
import numpy as np

plt.ion()

theta = np.linspace(0, 2 * np.pi, 200)

for t in np.linspace(0, 10, 100):
    plt.clf()

    for i in range(5):
        angle = t + i * 1.2
        r = 1 + 0.3 * np.sin(3 * theta + angle)
        x = r * np.cos(theta + angle)
        y = r * np.sin(theta + angle)
        plt.fill(x, y, alpha=0.5)

    plt.scatter(0, 0, s=300, color="cyan", alpha=0.3)
    plt.xlim(-2, 2)
    plt.ylim(-2, 2)
    plt.gca().set_facecolor("black")
    plt.axis("off")
    plt.title("💧 Liquid Orb Loader", color="white")
    plt.pause(0.05)

plt.ioff()
plt.show()
```
