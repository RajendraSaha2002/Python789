```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pyvista as pv

pv.OFF_SCREEN = False  # ensure desktop rendering

x, y, z = np.mgrid[-5:5:30j, -5:5:30j, 0:10:30j]

u = -y
v = x
w = 2 - z / 5.0

grid = pv.StructuredGrid(x, y, z)
grid["vectors"] = np.stack((u, v, w), axis=-1).reshape(-1, 3)

stream = grid.streamlines(vectors="vectors", n_points=200)

plotter = pv.Plotter()
plotter.set_background("black")
plotter.add_mesh(stream.tube(radius=0.03), color="cyan")

plotter.show()
```
