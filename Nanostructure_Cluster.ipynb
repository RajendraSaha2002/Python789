```python
import py3Dmol
import random
```

```python
# Create viewer
v = py3Dmol.view(width=500, height=500)

# Generate random nanocluster spheres
for _ in range(100):
    v.addSphere({
        "center": {
            "x": random.uniform(-2, 2),
            "y": random.uniform(-2, 2),
            "z": random.uniform(-2, 2)
        },
        "radius": 0.3,
        "color": "gold"
    })

# Styling
v.setBackgroundColor("black")
v.zoomTo()

# Optional rotation
v.spin(True)

v.show()
```
