```python
import py3Dmol
```

```python
xyz = """8
polymer
C 0.0 0.0 0.0
C 1.5 0.0 0.0
C 3.0 0.0 0.0
C 4.5 0.0 0.0
H 0.0 -1.0 0.0
H 1.5 1.0 0.0
H 3.0 -1.0 0.0
H 4.5 1.0 0.0
"""

v = py3Dmol.view(width=500, height=400)

v.addModel(xyz, "xyz")

v.setStyle({'stick': {}, 'sphere': {'scale': 0.3}})
v.zoomTo()

# Smooth rotation
v.spin(True)

v.show()
```
