```python
import pydeck as pdk
import pandas as pd
```

```python
data = pd.DataFrame({
    "city": ["Delhi", "Mumbai", "Chennai"],
    "lat": [28.6139, 19.0760, 13.0827],
    "lon": [77.2090, 72.8777, 80.2707],
    "height": [300000, 200000, 250000]
})
```

```python
layer = pdk.Layer(
    "ColumnLayer",
    data,
    get_position='[lon, lat]',
    get_elevation="height",
    elevation_scale=0.1,   # scaled for visibility
    radius=50000,
    get_fill_color=[0, 128, 255],
    pickable=True,
    auto_highlight=True
)
```

```python
view = pdk.ViewState(
    latitude=22,
    longitude=78,
    zoom=4,
    pitch=45
)

deck = pdk.Deck(
    layers=[layer],
    initial_view_state=view,
    tooltip={"text": "{city}\nHeight: {height}"}
)

deck
```
