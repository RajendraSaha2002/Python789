```python
import plotly.graph_objects as go
```

```python
cities = ["Delhi", "New York"]

lat = [28.6139, 40.7128]
lon = [77.2090, -74.0060]
```

```python
fig = go.Figure()

fig.add_trace(go.Scattergeo(
    lon=lon,
    lat=lat,
    text=cities,
    mode="markers+text",
    textposition="top center",
    marker=dict(
        size=10,
        color="red",
        line=dict(width=1, color="white")
    )
))

fig.update_geos(
    projection_type="orthographic",   # 3D globe effect
    showland=True,
    landcolor="rgb(200, 200, 200)",
    showocean=True,
    oceancolor="rgb(0, 100, 200)",
    showcountries=True,
    countrycolor="white"
)

fig.update_layout(
    title="🌍 Interactive 3D Globe",
    margin=dict(l=0, r=0, t=40, b=0)
)

fig.show()
```
