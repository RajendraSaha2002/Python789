```python
import folium
import random
from IPython.display import display
```

```python
# Base location (Delhi)
center_lat, center_lon = 28.6139, 77.2090

# Create map
m = folium.Map(location=[center_lat, center_lon], zoom_start=11)

# Add sensors
for i in range(20):
    val = random.randint(20, 100)

    lat = center_lat + random.uniform(-0.1, 0.1)
    lon = center_lon + random.uniform(-0.1, 0.1)

    color = "green" if val < 50 else "red"

    folium.CircleMarker(
        location=[lat, lon],
        radius=6,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.7,
        popup=f"📡 Sensor Value: {val}"
    ).add_to(m)

m
```

```python
m.save("smart_city_map.html")
print("✅ Map saved as smart_city_map.html")
```
