```python
import warnings
warnings.filterwarnings("ignore")

from vpython import *
from math import cos, sin

# Scene setup
scene.width = 900
scene.height = 600
scene.background = color.black
scene.title = "3D Solar System (Timed Run)"
scene.autoscale = False

# Objects
sun = sphere(pos=vector(0,0,0), radius=1, color=color.yellow, emissive=True)

earth = sphere(pos=vector(3,0,0), radius=0.3,
               color=color.blue, make_trail=True, retain=200)

mars = sphere(pos=vector(5,0,0), radius=0.25,
              color=color.red, make_trail=True, retain=200)

# Time control
t = 0
dt = 0.05

duration = 20      # seconds you want to run
fps = 60
steps = duration * fps

for i in range(int(steps)):
    rate(fps)

    earth.pos = vector(3*cos(t), 3*sin(t), 0)
    mars.pos = vector(5*cos(0.8*t), 5*sin(0.8*t), 0)

    t += dt

print("Simulation finished")
```
